# Example 1: Islet Storge access via S3 API
Islet storage service is available for each bridge within the DEDL.

Endpoint to access the storage serivce is:

_https://s3.{site_name}.data.destination-earth.eu_

site_name needs to be replaced with one ot the available bridges name.
- central
- lumi
- leonardo
- marenostrum
- eumetsat

## Access via Python (s3fs)

```python
import s3fs
from rich.prompt import Prompt

S3_BASE_URL = "https://s3.{site_name}.data.destination-earth.eu"

S3_URL_CENTRAL = S3_BASE_URL.format(site_name="central")

s3_central = s3fs.S3FileSystem(
    endpoint_url=S3_URL_CENTRAL,
    key=Prompt.ask(prompt="S3 Key"),
    secret=Prompt.ask(prompt="S3 Secret", password=True),
    use_ssl=True,
)
```

```python
s3_central.ls("")
```
